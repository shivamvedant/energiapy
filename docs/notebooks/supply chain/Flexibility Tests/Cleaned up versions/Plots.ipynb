```python
import matplotlib.pyplot as plt
import pickle
```

```python
MIP_gap = 0.05
smolyak_level = 4
quad_method = 'smolyak'
```

```python
load_combined_results = {}
eps_list = [0, 0.05, 0.1, 0.15, 0.2, 0.25, 0.3, 0.35, 0.4, 0.45, 0.5]

for eps in eps_list:
    
    with open(f"new_pkl_output\MIPGap_{int(MIP_gap*100):03d}\\results_{int(eps*100):03d}_{quad_method}_{smolyak_level}.pkl", 'rb') as file:
        load_results = pickle.load(file)
    
    load_combined_results[eps] = load_results
```

```python
load_combined_results[0.25]
```

```python
load_combined_results[0.3]
```

```python
load_combined_results[0.35]
```

```python
# Obj vs eps
# plot_df_gap = df_target.dropna(subset=["gap_pct"])

obj_list = [load_combined_results[j]['Obj'] for j in eps_list]

plt.figure(figsize=(7,5))
plt.plot(eps_list, obj_list, marker="o", ms=3)
plt.xlabel("ESF Target")
plt.ylabel("Objective Value")
# plt.xscale("log")
# plt.yscale("log")   # optional but often useful
plt.grid(True, alpha=0.3)
plt.title("Objective Values vs ESF Target")
plt.show()
```
